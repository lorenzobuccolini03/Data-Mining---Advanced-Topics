# Advanced Regression

Creating the file 'cmi_internet_regression.csv' which contains the same records in 'cmi_internet_cleaned.csv', except those that had a missing value in 'PCIAT-PCIAT_Total' (new column added at the end, target variable for regression).

```python
%matplotlib inline
import pandas as pd
```

```python
df0 = pd.read_csv('cmi_internet.csv')
df1 = pd.read_csv('cmi_internet_imputed.csv')
df2 = pd.read_csv('cmi_internet_cleaned.csv')
out = pd.read_csv('outliers_matrix.csv')
```

```python
# Create a new DataFrame df with all the columns of df1 + the column 'PCIAT-PCIAT_Total' from df0:
df = pd.concat([df1, df0['PCIAT-PCIAT_Total']], axis=1)

# Add a column in out with the mode of the values per row, and then add this column at the end of df:
out['mode'] = out.mode(axis=1)[0]
df = pd.concat([df, out['mode']], axis=1)
```

```python
# Check: count how many rows have 1 in the mode column:
df['mode'].value_counts()
```

```python
# Perfect, it matches the rows of internet_cleaned:
df2.shape
```

```python
# Remove all rows of df that have 1 in the mode column:
df = df[df['mode'] != 1]

# Remove all rows of df that have missing values in the column 'PCIAT-PCIAT_Total':
df = df.dropna(subset=['PCIAT-PCIAT_Total'])

# Remove the mode column from df:
df = df.drop(columns=['mode'])
```

```python
# Save df to a csv file called 'cmi_internet_regression.csv':
df.to_csv('cmi_internet_regression.csv', index=False) 
```
